### fonction qui prend une liste de temmperature et calcul la moyenne

```python
def moyenne(liste):
    resultat = sum(liste) / len(liste)

    return resultat

liste = [12, 25, 35, 25, 12, 19, -7, 12]
```

```python
moyenne(liste)
```
